```python
xt = tf.Variable(3.0)
with tf.GradientTape() as tape:
  yt = xt ** 2
print("TensorFlow dy/dx at x=3 :", tape.gradient(yt, xt).numpy())
```

```python
xp = torch.tensor(3.0, requires_grad=True)
yp = xp ** 2
yp.backward()
print("PyTorch dy/dx at x=3 :", xp.grad.item())
```
